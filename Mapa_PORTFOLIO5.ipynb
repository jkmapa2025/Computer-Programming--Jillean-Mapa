```python
### Introduction ###
# FEATURE: .center() perfectly aligns the text in the middle of the 60 equal signs
print(f"\n{'='*60}")
print("Welcome to Hotel Marina!".center(60))
print(f"{'='*60}")

### Stored data for occupied and available rooms and their prices ###
room_prices = {101: 1000, 102: 1000, 103: 1000, 104: 1000, 105: 1000, 110: 1500, 111: 1500, 120: 2000, 130: 3000}
occupiedrooms = [101, 103, 104]
availrooms = [102, 105, 110, 111, 120, 130]

### Storing of the Name of the Guest ###
### .title() ensures names like "jillean mapa" become "Jillean Mapa" ###
name = input("Name of guest: ").strip().title()
print("\n" + f"Welcome Ma'am/Sir {name}, to Hotel Marina!".center(60) + "\n")

### Initial Values ### 
total_roomprice = 0
total_services = 0
days = 0
room = 0

### A loop for the room reservation ###
while True:
    print(f"Available Rooms are {availrooms} \n{'_'*60}")

    try:
        room = int(input("Room No: "))
    except ValueError:
        print("Invalid input. Please enter a valid room number.")
        continue

    if room in availrooms:
        try:
            days = int(input("Days of Stay: "))
            if days <= 0:
                print("Days must be 1 or more.")
                continue

            total_roomprice = room_prices[room] * days
            print(f"\n{'_'*70}")
            print(f"Success! Room {room} booked for {days} day(s). Room total: Php {total_roomprice}")
            print(f"{'_'*70}\n")
            break 
        except ValueError:
            print("Invalid input. Please enter a number for days.")
    else:
        print("Room not available or does not exist. Please try again.\n")

### Offers the Order and stores it ###
###.startswith('y') allows the user to type "y", "yes", "yeah", etc. ###
orders = input("Additional Services (Requests, Orders,...etc.)? (yes/no): ").strip().lower()

if orders.startswith('y'):
    while True:
        print("\nAdditional Services include Drinks(A), Food(B) and other requests (C)")
        service = input("Select Service (A/B/C) or type 'CANCEL' to finish: ").strip().upper()

        if service == "CANCEL":
            break

        elif service == "A":
            print("Drinks: (1) Alcoholic= Php 130, (2) Non-Alcoholic= Php 100")
            try:
                drink_order = int(input("Order (Enter 1 or 2): "))
                quantity = int(input("Quantity: "))
            except ValueError:
                print("Invalid input. Please use numbers.")
                continue
                
            if drink_order == 1:
                drink_name = "Alcoholic Drink"
                price = 130 * quantity
            elif drink_order == 2:
                drink_name = "Non-Alcoholic Drink"
                price = 100 * quantity
            else:
                print("Invalid drink choice.")
                continue 

            total_services += price
            print(f"{'-'*60}\n Added {quantity}x {drink_name} | Subtotal: Php {price} \n{'-'*60}")

        elif service == "B":
            print("Food: (1) Breakfast= Php 400, (2) Lunch= Php 650, (3) Dinner= Php 800")
            try:
                food_order = int(input("Order (Enter 1-3): "))
                quantity = int(input("Quantity: "))
            except ValueError:
                print("Invalid input. Please use numbers.")
                continue

            if food_order == 1:
                food_name = "Breakfast Bundle"
                price = 400 * quantity
            elif food_order == 2:
                food_name = "Lunch Bundle"
                price = 650 * quantity
            elif food_order == 3:
                food_name = "Dinner Bundle"
                price = 800 * quantity
            else:
                print("Invalid food choice.")
                continue

            total_services += price
            print(f"{'-'*60}\n Added {quantity}x {food_name} | Subtotal: Php {price} \n{'-'*60}")
            
        elif service == "C":
            print("Other requests are not yet available.")
            continue
            
        else:
            print("Invalid selection. Please choose A, B, or CANCEL.")
            continue

        more = input("Would you like to order more? (yes/no): ").strip().lower()
        if not more.startswith('y'):
            print("Thank you! Proceeding to checkout...")
            break

### Promo Code Section: The promo code is "VIP" ###
final_bill = total_roomprice + total_services
discount = 0

print("\n" + "="*60)
promo = input("Enter promo code (or press Enter to skip): ").strip().upper()
if promo.startswith('VIP'):
    print("VIP Promo applied! 10% discount on total bill.")
    discount = final_bill * 0.10
    final_bill -= discount
elif len(promo) > 0:
    print("Invalid promo code.")

### The Final Receipt ###
### Using .ljust() (left justify) and .rjust() (right justify) to create proper receipt columns ###
print("\n" + "=" * 60)
print("HOTEL MARINA - OFFICIAL RECEIPT".center(60))
print("=" * 60)

print(f"Guest Name: {name}")
print("-" * 60)

### Formats the strings to take up specific amounts of space so the prices align perfectly on the right ###
print("Room Booking".ljust(40) + f"Php {total_roomprice:,.2f}".rjust(20))

if total_services > 0:
    print("Additional Services".ljust(40) + f"Php {total_services:,.2f}".rjust(20))

if discount > 0:
    print("Discount Applied".ljust(40) + f"-Php {discount:,.2f}".rjust(20))

print("-" * 60)
print("TOTAL AMOUNT DUE".ljust(40) + f"Php {final_bill:,.2f}".rjust(20))
print("=" * 60)
print("Thank you for choosing Hotel Marina!".center(60))
print("=" * 60)
```
