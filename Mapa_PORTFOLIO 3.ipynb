```python
### Calculates the Total Price ###
def meat_price(weight, price_per_kg):
    total = weight * price_per_kg
    return total

### Rate of Meat per Kilogram ###
beef_rate = 140.0
pork_rate = 80.0
chicken_rate = 90.0
lamb_rate = 150.0

print("Welcome to All MEAT Meat Shop!")

### Visual Aid for Available Meats ###
print(f"\n{'='*60}\n AVAILABLE MEATS!\n{'='*60}")
print(f" (A) BEEF {"-"*30} Php {beef_rate}/kg")
print(f" (B) PORK {"-"*30} Php {pork_rate}/kg")
print(f" (C) CHICKEN {"-"*27} Php {chicken_rate}/kg")
print(f" (D) LAMB {"-"*30} Php {lamb_rate}/kg")

### Order Input of the Customer ###
print("  "*30)
order = input("What meat would you like to order? ") 
weight = float(input("How many kilograms?"))

### Calculates the Meat Price ###
if order == 'A' or 'BEEF':
    total_meatprice = meat_price(weight, beef_rate)
elif order == 'B' or 'PORK':
    total_meatprice = meat_price(weight, pork_rate)
elif order == 'C' or 'CHICKEN':
    total_meatprice = meat_price(weight, chicken_rate)
elif order == 'D' or 'LAMB':
    total_meatprice = meat_price(weight, lamb_rate)
else:
    total_meatprice = None 
    print("="*30)
    print("Sorry, we don't have that meat in stock.")

### if there is an order, this is the receipt ###
if total_meatprice is not None:
    print(f"\n{'='*30}\n Your total for {weight}kg of {order} is: Php {total_meatprice}")
```
