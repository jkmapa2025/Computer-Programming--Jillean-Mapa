#### Hotel and Ordering Service 

```python
# Introduction ###
print(f"\n{'='*60}\n Welcome to Hotel Marina! \n{'='*60}")

### Stored data for occupied and available rooms and it's prices ###
room_prices = {101: 1000, 102: 1000, 103: 1000, 104: 1000, 105: 1000, 110: 1500, 111: 1500, 120: 2000, 130: 3000}
occupiedrooms = [101, 103, 104]
availrooms = [102, 105, 110, 111, 120, 130]

### Storing of the Name of the Guest ###
name = input("Name of guest: ")
print("  "*30)
print(f"Welcome Ma'am/Sir {name}, to Hotel Marina!")

### Initial Values of the price of the room and services before the input of order ### 
total_roomprice = 0
total_services = 0
days = 0
room = 0

### A loop for the room reservation ###
while True:
    print(f"\n Available Rooms are {availrooms} \n{'_'*60}")

    ### Value Error, the room number that was entered is not valid ###
    try:
        room = int(input("Room No: "))
    except ValueError:
        print("Invalid input. Please enter a valid room number.")
        continue

    ### If the room is part of the available rooms given ealier, then it calculates the room price ###
    if room in availrooms:
        try:
            days = int(input("Days of Stay: "))
            if days <= 0:
                print("Days must be 1 or more.")
                continue

            ### The room price is calculated from the stored data from earlier with this formula ###
            total_roomprice = room_prices[room] * days
            print(f"\n{'_'*90}\n You have successfully booked with Hotel Marina! Room total is Php {total_roomprice} for Ma'am/Sir {name} \n{'_'*90}")
            break 
        except ValueError:
            print("Invalid input. Please enter a number for days.")
    else:
        print("Room not available or does not exist. Please try again.")

### Offers the Order and stores it ###
print("  "*30)
orders = input("Additional Services (Requests, Orders,...etc.)? yes or no? ").strip().lower()

### If yes, then it gives the following options for the user ###
if orders != "no":
   ### It is under a loop where if the order is CANCELED then it will proceed to the final bill ###
    while True:
        print("Additional Services include Drinks(A), Food(B) and other requests (C)")
        service = input("Select Service (A/B/C) or 'CANCEL' to finish ordering: ").strip().upper()

        if service == "CANCEL":
            break

        ### If the user chose A, it will give the folloing choices ###
        elif service == "A":
            print("Drinks: (1) Alcoholic= 130, (2) Non-Alcoholic= 100")
            try:
                drink_order = int(input("Order (Enter number 1 or 2): "))
                quantity = int(input("Quantity: "))
            except ValueError:
                print("Invalid input. Please use numbers.")
                continue
                
            ### The price of each option in drinks ###
            if drink_order == 1:
                drink_name = "Alcoholic Drink"
                price = 130 * quantity
            elif drink_order == 2:
                drink_name = "Non-Alcoholic Drink"
                price = 100 * quantity
            else:
                print("Invalid drink choice.")
                continue 

            ### If there is an order, the price of that order will add to the total services ordered by the user ###
            total_services += price
            print(f"\n{'_'*60}\n You added {quantity} {drink_name:<20} amounting to Php {price} \n{'_'*60}")
            print("  "*35)

        ### Similar with service A ###
        elif service == "B":
            print("(1) Breakfast Bundle= 400, (2) Lunch Bundle= 650, (3) Dinner Bundle= 800")
            try:
                food_order = int(input("Order (Enter number 1-3): "))
                quantity = int(input("Quantity: "))
            except ValueError:
                print("Invalid input. Please use numbers.")
                continue

            ### The Price of the Food Order ###
            if food_order == 1:
                food_name = "Breakfast Bundle"
                price = 400 * quantity
            elif food_order == 2:
                food_name = "Lunch Bundle"
                price = 650 * quantity
            elif food_order == 3:
                food_name = "Dinner Bundle"
                price = 800 * quantity
            else:
                print("Invalid food choice.")
                continue

            ### If there is an order, then it will add to the total services ordered ###
            total_services += price
            print(f"\n{'_'*60}\n You added {quantity} {food_name:<20} amounting to Php {price} \n{'_'*60}")
            print("  "*35)
            
        elif service == "C":
            print("Other requests are not yet available.")
            print("  "*35)
            continue
            
        else:
            print("Invalid selection. Please choose A, B, or CANCEL.")
            continue

        ### If the user wants to order more, it will loop back to the choices of services ###
        more = input("Would you like to order more? (yes/no): ").strip().lower()
        ### If there is no order, then it will make the final reciept ###
        if more == "no":
            print("Thank you very much.")
            break

### The final bill is calculated with the room price and the total services bought by the user ###
final_bill = total_roomprice + total_services

### The final Reciept ###
print("  "*35)
print("==" * 50)
print(f"Ma'am/Sir {name} booked {days} days for room no. {room} amounting to {' '*8} Php {total_roomprice}")
if total_services > 0:
    print(f"Ma'am/Sir {name} ordered services amounting to {' '*22} Php {total_services}")
print("==" * 50)
print(f"Final Amount for Ma'am/Sir {name:<40} Php {final_bill}")
print("Thank you for choosing Hotel Marina!")
```
