### Updated All Meat Meat Shop

```python
### Calculates the Total Price ###
def meat_price(weight, price_per_kg):
    return weight * price_per_kg

### Inventory of available meats in the shop ###
meat_codes = ['A', 'B', 'C', 'D']
meat_types = ['BEEF', 'PORK', 'CHICKEN', 'LAMB']
meat_rates = [140.0, 80.0, 90.0, 150.0]

### Active Promo Codes, Try to Input the following: "MEAT10", "SUPERSTEAK", & "FREESHIP" ###
PROMO_CODES = {
    "MEAT10": 0.10,
    "SUPERSTEAK": 0.15,
    "FREESHIP": 0.00
}

print("Welcome to All MEAT Meat Shop!")

### New and Improved Visual Aid for Available Meats ###
print(f"\n{'='*60}\nAVAILABLE MEATS!\n{'='*60}")
for i in range(len(meat_types)):
    print(f"({meat_codes[i]}) {meat_types[i]} {'-'*30} Php {meat_rates[i]}/kg")

### Tracks the input of the customer ###
cart_items = []
grand_total = 0.0

### Continuous loop until the customer is done ordering ###
while True:
    print("\n" + "*" * 30)
    order = input("What meat would you like to order? (or type 'DONE' to checkout): ").strip().upper()
    
    ### to break out of the loop ###
    if order == 'DONE':
        break
        
    # Check if the input is valid using list menu
    if order not in meat_codes and order not in meat_types:
        print("Sorry, we don't have that meat in stock. Please try again.")
        continue  # Skips the rest of the loop iteration and asks again
        
    ### Find the corresponding index in our inventory ###
    if order in meat_codes:
        index = meat_codes.index(order)
    else:
        index = meat_types.index(order)
        
    meat_name = meat_types[index]
    rate = meat_rates[index]
    
    ### validation for weight ###
    try:
        weight = float(input(f"How many kilograms of {meat_name}? "))
        if weight <= 0:
            print("Please enter a weight greater than 0.")
            continue
    except ValueError:
        print("Invalid weight. Please enter a numerical value.")
        continue
        
    # Calculate item price and append to our tracking systems
    item_total = meat_price(weight, rate)
    grand_total += item_total
    cart_items.append(f"{weight}kg of {meat_name} (Php {item_total:.2f})")
    print(f"-> Added to cart: {weight}kg of {meat_name}")

###Promo Code System ===
discount_amount = 0.0

if grand_total > 0:
    print("\n" + "=" * 30)
    promo_input = input("Enter a promo code (or press Enter to skip): ").strip().upper()
    
    if promo_input in PROMO_CODES:
        discount_rate = PROMO_CODES[promo_input]
        discount_amount = grand_total * discount_rate
        print(f"Success! Promo code '{promo_input}' applied ({discount_rate * 100}% off).")
    elif promo_input != "":
        print("Invalid promo code. Proceeding without discount.")

### Final Receipt Generation ###
if grand_total > 0:
    print(f"\n{'='*60}\nFINAL RECEIPT\n{'='*60}")
    for item in cart_items:
        print(f" - {item}")
    print("-" * 60)
    print(f"Subtotal: {' ' * 31} Php {grand_total:.2f}")
    
    if discount_amount > 0:
        print(f"Discount: {' ' * 31}-Php {discount_amount:.2f}")
        
    final_total = grand_total - discount_amount
    print(f"Total Amount Due: {' ' * 23} Php {final_total:.2f}")
    print('='*60)
else:
    print("\nNo items were added to your cart. Have a wonderful day!")
```
