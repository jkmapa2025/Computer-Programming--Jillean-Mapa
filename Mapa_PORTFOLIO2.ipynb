#### EMailIt: Letter and Package Sending Service ####

```python
### Initial Values ###
mailfee = 30
initial_package_fee = 39

### Name Input for Sender ###
name = input("Enter name of sender: ")
print(f"Welcome to EMailIt, {name}\n")

### For the Recipient ###
to = input("Enter name of recipient: ")
location = input("Enter the location of sender: ")
where = input("Enter the location of the recipient: ")
message = input("Enter message: ")

print("__" * 35)

### Package Option ###
has_package = input("Do you have a package to send? (yes/no): ")

### If yes to the package ### 
if has_package != "no":
    print("Awesome! Let's go to the specifics.")
    print(f"The delivery fee is {initial_package_fee} with an added 2/kilo for weight over 20kg.")
    
    weight = float(input("Enter the weight of the package (kg): "))
    is_fragile = input("Is the package fragile? (yes/no): ")

    ### Charge for Package ###
    if weight <= 20:
        total_package_fee = initial_package_fee

    ### Charge for Package over 20kg ###
    else:
        extra_weight = weight - 20
        total_package_fee = initial_package_fee + (extra_weight * 2)

    overall_amount = total_package_fee + mailfee

    ### The Final Receipt with Package ###
    print("*" * 35)
    print(f"The mail fee: {mailfee}")
    print(f"The total package fee: {total_package_fee:.2f}")
    print(f"OVERALL AMOUNT: {overall_amount:.2f}, {name}!")
    print("*" * 35)

### Final Receipt (Mail Only) ###
else:
    print("\n" + "*" * 35)
    print(f"Noted! Your total is {mailfee}, {name}!")
    print("*" * 35)

    print("Thank you for using EMailIt!!")
```
